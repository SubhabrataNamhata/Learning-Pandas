```python
#Lets make a general function to find again which theam a certain plr scored most runs.

def run_scored(batsman_name):
  player= ipl[ipl['batsman'] == batsman_name]
  return player.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).head(1)
```

```python
batsman_name = input("Enter the batsman name : ")
run_scored(batsman_name)
```

```python
# isin()

# Find the most destructive death over batsman in the history of IPL

# Strike Rate= (Number of runs/Number of balls)/100

# Min batsman 200 balls in over 16-20
```

```python
mask1 = ipl['over']>15
death_over = ipl[mask1]
allbatsman = death_over.groupby('batsman')['batsman_runs'].count()
mask2 = allbatsman>200
Top20_destructive_batsmans= allbatsman[mask2].sort_values(ascending=False).head(20)
Top20_destructive_batsmans

```

```python

Top20_destructive_batsmans.index.tolist()
```

```python
# Runs scored and balls playedby these 20 batsmans
listed = Top20_destructive_batsmans.index.tolist()
mask = ipl['batsman'].isin(listed)
final=ipl[mask]
```

```python
runs=final.groupby('batsman')['batsman_runs'].sum().sort_values(ascending=False)
balls=final.groupby('batsman')['batsman_runs'].count().sort_values(ascending=False)
sr=(runs/balls)*100
#sr
#runs
balls

```

###Pivot Table

###Filling Missing Values using fillna()

```python
#missing.fillna("NA")
missing['Stock'].fillna(method='bfill')#Copies previous value downward.
#method=ffill Copies next value upwards
#Thre are many features like these


# replace missing ratings with the mean of available ratings.
missing['Rating'].fillna(missing['Rating'].mean())
```
